```python
# 14. Write a Python program to check whether a year is a leap year.
y = int(input("enter the year "))
leap = False
if y % 4 == 0 :
    if y %100 :
        if y%400 == 0 :
            leap = True
        leap = False 
    leap = True
print(leap)
```

```python
# 16. Write a Python program to generate the Fibonacci series up to n terms.
n = int(input("enter a number "))
lis = [0,1]
count = n-1
while count != 0 :
    lis.append(lis[-2]+lis[-1])
    count -= 1 
print(lis)
```

```python
# 26. Write a Python program to find the GCD of two numbers.
a = int(input("enter a number"))
b = int(input("enter another number "))
alis = []
blis = []
clis = []
for i in range (1,a+1):
    if a%i == 0 :
        alis.append(i)
for i in range (1,b+1):
    if b%i == 0 :
        blis.append(i)
for i in alis :
    if i in blis:
        clis.append(i)
max(clis)
```

```python
# 27. Write a Python program to find the LCM of two numbers.
a = int(input("enter a number"))
b = int(input("enter another number "))
alis = []
blis = []
clis = []
for i in range (1,a+1):
    if a%i == 0 :
        alis.append(i)
for i in range (1,b+1):
    if b%i == 0 :
        blis.append(i)
for i in alis :
    if i in blis:
        clis.append(i)
print((a*b)/max(clis))
```
