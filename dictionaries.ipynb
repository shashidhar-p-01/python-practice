```python
# Create a dictionary containing student name and age.
d = {"name":"appu" , "age":23}
print(d)
```

```python
# Print all keys of a dictionary.
print(d.keys())
```

```python
# Print all values of a dictionary.
print(d.values())
```

```python
# Access a value using its key.
print(d["name"])
```

```python
# Add a new key-value pair to a dictionary.
print(d)
d["new_key"] = "new_value"
print(d)
```

```python
# Update the value of an existing key.
print(d)
d["new_key"] = "update"
print(d)
```

```python
# Remove a key from a dictionary using pop().
d.pop("new_key")
print(d)
```

```python
# Remove the last inserted item using popitem().
d.popitem()
print(d)
```

```python
# Check whether a key exists in a dictionary.
print("name" in d.keys())
```

```python
# Find the length of a dictionary.
print(len(d))
```

```python
# Print all key-value pairs using a loop.
for keys,values in d.items():
    print(f"{keys} : {values}")
```

```python
# Create a dictionary from two lists (keys and values).
keys = ["name" , "age" , "city"]
values = ["appu" , 23 , "banglore"]
d = {}
for i in range(len(keys)):
    d[keys[i]] = values[i]
print(d)
```

```python
# Get all keys using keys().
d = {'name': 'appu', 'age': 23, 'new_key': 'new_value'}
print(d.keys())
```

```python
# Get all values using values().
print(d.values())
```

```python
# Get all key-value pairs using items().
print(d.items())
```

```python
# Create a nested dictionary for student details.
d = {"name" : "appu" , "age" : 20 , "marks" : {"sub1" : 40 , "sub2" : 50 , "sub3" : 60}}
print(d)
```

```python
# Find the maximum value in a dictionary.
d = {"sub1" : 40 , "sub2" : 50 , "sub3" : 60}
a = d.values()
print(a)
print(max(a))
```

```python
# Find the minimum value in a dictionary.
print(min(a))
```

```python
# Merge two dictionaries.
d = {"sub1" : 40 , "sub2" : 50 , "sub3" : 60}
d2 = {"name":"appu" , "age":23}
d.update(d2)
print(d)
```

```python
# Count the frequency of each character in a string using a dictionary.
s = "this is a string "
d = {}
for char in s :
    if char not in d:
        d[char] = s.count(char)
print(d)
```
