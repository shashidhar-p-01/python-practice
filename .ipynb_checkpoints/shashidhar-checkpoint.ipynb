```python
# 1. Print fibonacci series
lis = [0,1]
for i in range(0,10):
    lis.append(lis[i]+lis[i+1])
print(lis)
```

```python
# 2. Find first non-repeating character 
lis = list("appu was here")
unique = {}
for char in lis:
    if char in unique:
        pass
    else :
        unique[char] = lis.count(char)
for key,value in unique.items():
    if value == 1:
        print(f"non-repeating first character is {key}")
        break 
```

```python
# 3. Remove duplicate elements from a list 
lis = ['a','a','p','p','u']
lis2 = set(lis)
print(lis2)
```

```python
# 4. print floyd's triangle 
n = 15 
num = 1
count = 0 
for i in range(1, n+1):
    for j in range(i):
        if count<n:
            print(num, end=" ")
            num += 1
            count+=1
        else:
            break
    if count>=n:
        break 
    print()
```

```python
# 5. Create simple calculator 
num1 = int(input("enter a number"))
num2 = int(input("enter another number"))
oper = input("enter the operator + , - , * , /")
match oper :
    case "+":
        print(num1+num2)
    case "-":
        print(num1-num2)
    case "*":
        print(num1*num2)
    case "/":
        print(num1/num2)
```

```python
# 6. Count digits 
lis = [n for n in range(0,10)]
sum = 0 
for i in lis :
    sum = sum+i
print(sum)
```

```python
# 7. find pair with given sum 
lis = [n for n in range(0,11)]
lis2 = []
for i in range(0,len(lis),2):
    if i+1 < len(lis):
        lis2.append(lis[i]+lis[i+1])
print(lis2)
```

```python
# 8. Check leap year 
year = int(input("Enter an year "))
leap = True
if year % 4 == 0 or year %100 == 0 :
    pass
else :
    leap = False 
print(leap )
```

```python
# 9. Check pallindrome using recusrion 
def pallindrome(a):
    rev = a[::-1]
    if a == rev:
        print ("is pallindrome")
    else:
        print("not pallindrome")
a = input("enter a text")
pallindrome(a)
```

```python
# 10. Convert decimal to binary 
lis = [0,1.3,2.4,9,0,3.4,8,5]
lis2 = []
for i in lis:
    if type(i) == float:
        lis2.append(0)
    else :
        lis2.append(i)
print(lis2)
```
