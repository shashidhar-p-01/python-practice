```python
# fint the frequency of each character in a string 
a = "this is a string"
d = {}
for char in a :
    if char not in d :
        d[char] = a.count(char)
print(d)
    
```

```python
# find the longest word in a sentance 
a = "this is a sentance "
l = a.split()
longest = ""
print(l)
for i in l :
    if len(i) > len(longest):
        longest = i 
print(longest)
```

```python
# check weather 2 strings are anagrams 
s1 = "this is a string"
s2 = "a string is this"
d = {}
for char in s1:
    if char not in d:
        d[char] = s1.count(char)
d2 = {}
for char in s2:
    if char not in d2:
        d2[char] = s2.count(char)
print(d == d2)
```
