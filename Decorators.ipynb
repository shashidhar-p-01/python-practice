```python
# Create decorator to print welcome message before function execution.
def dec_fun(func):
    def wrapper():
        print("welcome")
        func()
    return wrapper
@dec_fun
def fun():
    print("appu")
fun()
```

```python
# Create decorator to print before and after function execution.
def dec_fun(func):
    def wrapper():
        print("welcome")
        func()
        print("have a good day")
    return wrapper
@dec_fun
def fun():
    print("appu")
fun()
```

```python
# Create decorator to check login before dashboard access.
def dec_fun(func):
    def wrapper():
        if "appu@gmail.com" == input("enter mail"):
            if "APPUgowda" == input("enter password"):
                func()
            else:
                print("wrong password")
        else:
            print("wrong email")
        print("have a good day")
    return wrapper
@dec_fun
def fun():
    print("login successfull")
fun()
```

```python
# Create decorator to calculate function execution time.
import time 
def dec_fun(func):
    def wrapper():
        start = time.time()
        func()
        end = time.time()
        print(f"{end-start}")
    return wrapper
@dec_fun
def fun():
    for i in range(1000000):
        pass
fun()
```
