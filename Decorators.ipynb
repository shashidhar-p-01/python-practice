```python
# Create decorator to print welcome message before function execution.
def dec_fun(func):
    def wrapper():
        print("welcome")
        func()
    return wrapper
@dec_fun
def fun():
    print("appu")
fun()
```

```python
# Create decorator to print before and after function execution.
def dec_fun(func):
    def wrapper():
        print("welcome")
        func()
        print("have a good day")
    return wrapper
@dec_fun
def fun():
    print("appu")
fun()
```

```python
# Create decorator to check login before dashboard access.
def dec_fun(func):
    def wrapper():
        if "appu@gmail.com" == input("enter mail"):
            if "APPUgowda" == input("enter password"):
                func()
            else:
                print("wrong password")
        else:
            print("wrong email")
        print("have a good day")
    return wrapper
@dec_fun
def fun():
    print("login successfull")
fun()
```

```python
# Create decorator to calculate function execution time.
import time 
def dec_fun(func):
    def wrapper():
        start = time.time()
        func()
        end = time.time()
        print(f"{end-start}")
    return wrapper
@dec_fun
def fun():
    for i in range(1000000):
        pass
fun()
```

```python
# Create decorator to convert output into uppercase.
def dec_fun(func):
    def wrapper(*args,**kwargs):
        result = func(*args,**kwargs)
        return result.upper()
    return wrapper
@dec_fun
def greet():
    return "hello world"
greet()
```

```python
# Create decorator to multiply returned result by 2.
def dec_fun(func):
    def wrapper(*args,**kwargs):
        result = func(*args,**kwargs)
        return result*2
    return wrapper
@dec_fun
def greet(a):
    return a
greet(6)
```

```python
# Create decorator to check integer input before calculation.
def dec_fun(func):
    def wrapper(*args):
        if func(*args).isdigit and int(func(*args)):
            print(f"{args} is an integer ")
    return wrapper
@dec_fun
def fun(a):
    return a
fun("78")
```
