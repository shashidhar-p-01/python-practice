```python
# 2. Create a Rectangle Class

# Create a class Rectangle with:

# length
# width

# Methods:

# area()
# perimeter()

class Rectangle:
    def __init__(self,length,width):
        self.length = length
        self.width = width

    def area(self):
        area = self.length * self.width
        print(area)

    def perimeter(self):
        peri = (self.length * 2) + (self.width * 2)
        print(peri)
rect1 = Rectangle(3,4)
rect1.area()
rect1.perimeter()
```

```python
# 3. Bank Account

# Create a class BankAccount with:

# account_holder
# balance

# Methods:

# deposit(amount)
# withdraw(amount)
# check_balance()

class BankAccount:
    def __init__(self,acc_holder,balance=0):
        self.acc_holder = acc_holder
        self.balance = balance

        
    def deposit(self,amount):
        self.balance += amount

        
    def withdraw(self,amount):
        self.balance -= amount

    
    def check_balance(self):
        print(self.balance)

user1 = BankAccount("Appu",90000)
user1.deposit(200)
user1.withdraw(5000)
user1.check_balance()
```

```python
# 7. Method Overriding

# Create:

# Animal

# with method:

# sound()

# Create:

# Dog
# Cat

# Override sound() in both classes.


class Animal:
    def __init__(self):
        print("Animal object created ")

    def sound(self):
        print("grr")

class Cat(Animal):
    def sound(self):
        print("meow")
class Dog(Animal):
    def sound(self):
        print("bark")

dog = Dog()
cat = Cat()
dog.sound()
cat.sound()
```

```python
# 8. Multilevel Inheritance

# Create:

# Person

# ↓

# Employee

# ↓

# Manager

# Display details from all classes.

class Person:
    def __init__(self):
        print("Im a person")

class Employee(Person):
    def __init__(self):
        super().__init__()
        print("An employee is a person")

class Manager(Employee):
    def __init__(self):
        super().__init__()
        print("A manager is an employee amd also a person ")

employee = Manager()
employee2 = Employee()
```

```python
# 9. IS-A Relationship

# Create:

# Person

# and

# Teacher

# Demonstrate that Teacher IS-A Person.

class Person:
    def __init__(self):
        print("is a Person")
    def feature(self):
        print("has eyes, legs , hands , nose ")

class Teacher(Person):
    def __init__(self):
        super().__init__()
        print("a teacher is a person")
        self.feature()
teach = Teacher()
```

```python
# 12. Getter and Setter

# Create a class Student.

# Store marks as private variable.

# Create:

# get_marks()
# set_marks()

class Student:
    def __init__(self,name,marks):
        self.name = name
        self.__marks = marks
    def get_marks(self):
        return self.__marks
    def set_marks(self,new):
        self.__marks = new
        print(f"updated marks{self.__marks}")

stu = Student("appu",25)
print(stu.get_marks())
stu.set_marks(30)
```

```python
# 13. Abstract Class

# Create an abstract class:

# Shape

# Abstract method:

# area()

# Create:

# Circle
# Square

# Implement area calculation.

from abc import ABC,abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def area(self,radius=3):
        print(3.172 * (radius*radius))

class Square(Shape):
    def area(self,l=10):
        print(l**2)
shape = Circle()
shape2 = Square()
shape.area()
shape2.area()
```
