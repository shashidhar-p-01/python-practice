```python
# Check whether a year is a leap year.
y = 2020
leap = False
if y%4 ==0 :
    if y%100 != 0 and y%400 == 0 :
        leap = True
    leap = True
print(leap)
```

```python
# Calculate grade: 90+ → A , 75–89 → B , 50–74 → C , Below 50 → Fail
m = 90
if m < 50:
    print("Fail")
elif 50 <= m <= 74:
    print("C")
elif 75 <= m <= 89:
    print("B")
elif m >= 90:
    print("A")
```

```python
# Build a simple calculator using if-else (+, -, *, /).
n1 = 3
n2 = 2
op = "+"
if op == "+":
    print(n1+n2)
elif op == "-":
    print(n1-n2)
elif op == "*":
    print(n1*n2)
elif op == "/":
    print(n1/n2)
```
